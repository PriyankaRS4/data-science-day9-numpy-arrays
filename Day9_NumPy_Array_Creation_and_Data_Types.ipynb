# NumPy Array Creation and Data Types

**Dataset:** Student Marks Dataset

This notebook demonstrates one-dimensional and two-dimensional NumPy arrays, array attributes, numerical operations, and a comparison between NumPy arrays and Python lists.

## Objective
Learn the fundamentals of NumPy, a core Python library for data science.

```python
import numpy as np
import pandas as pd

df = pd.read_csv('student_marks.csv')
display(df.head())
print('Dataset shape:', df.shape)
```

## 1. One-dimensional array

```python
marks_1d = np.array([78, 85, 92, 67, 88])
print('1D array:', marks_1d)
print('Shape:', marks_1d.shape)
print('Dimensions (ndim):', marks_1d.ndim)
print('Size:', marks_1d.size)
print('Data type (dtype):', marks_1d.dtype)
```

## 2. Two-dimensional array

```python
marks_2d = np.array([
    [78, 82, 75],
    [90, 88, 92],
    [65, 70, 68]
])
print('2D array:\n', marks_2d)
print('Shape:', marks_2d.shape)
print('Dimensions (ndim):', marks_2d.ndim)
print('Size:', marks_2d.size)
print('Data type (dtype):', marks_2d.dtype)
```

## 3. Array attributes: shape, ndim, size, dtype

```python
print('shape =', marks_2d.shape)
print('ndim  =', marks_2d.ndim)
print('size  =', marks_2d.size)
print('dtype =', marks_2d.dtype)
```

## 4. Numerical operations

```python
marks = np.array([60, 70, 80, 90, 100])
print('Original:', marks)
print('Add 5:', marks + 5)
print('Subtract 5:', marks - 5)
print('Multiply by 2:', marks * 2)
print('Divide by 2:', marks / 2)
print('Mean:', marks.mean())
print('Maximum:', marks.max())
print('Minimum:', marks.min())
print('Total:', marks.sum())
```

## 5. NumPy arrays using the Student Marks Dataset

```python
subject_columns = ['Maths', 'Physics', 'Chemistry', 'English', 'Biology', 'Economics', 'History', 'Civics']
student_marks = df[subject_columns].to_numpy(dtype=float)

print('Student marks array shape:', student_marks.shape)
print('ndim:', student_marks.ndim)
print('size:', student_marks.size)
print('dtype:', student_marks.dtype)
print('First student marks:', student_marks[0])
```

## 6. Numerical operations on student marks

```python
print('Average mark per subject:')
print(student_marks.mean(axis=0))

print('\nAverage mark per student:')
print(student_marks.mean(axis=1))

print('\nHighest mark in each subject:')
print(student_marks.max(axis=0))

print('\nLowest mark in each subject:')
print(student_marks.min(axis=0))
```

## 7. Compare NumPy arrays with Python lists

```python
python_list = [10, 20, 30, 40]
numpy_array = np.array([10, 20, 30, 40])

print('Python list * 2:', python_list * 2)
print('NumPy array * 2:', numpy_array * 2)
print('\nNumPy array + 5:', numpy_array + 5)

print('\nNumPy arrays support fast element-wise numerical operations, while Python lists are general-purpose collections.')
```

## Key Takeaways
- `np.array()` creates NumPy arrays.
- A 1D array has one axis; a 2D array has rows and columns.
- `shape` gives the dimensions, `ndim` gives the number of dimensions, `size` gives total elements, and `dtype` gives the data type.
- NumPy supports fast element-wise numerical operations.
- NumPy arrays are especially useful for numerical and data-science computations.

## Interview Questions
**1. What is NumPy?** — NumPy is a Python library for numerical computing that provides efficient multidimensional arrays and mathematical operations.

**2. What is the difference between a NumPy array and a Python list?** — NumPy arrays are optimized for numerical operations and support fast element-wise calculations; Python lists are more general-purpose.

**3. What does the shape attribute represent?** — `shape` tells us the size of the array along each dimension. For example, `(3, 4)` means 3 rows and 4 columns.
